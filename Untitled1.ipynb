```python
from ultralytics import YOLO

print("Ultralytics YOLO installed successfully")
```

```python
data_yaml = """
path: /content/data
train: train/images
val: validation/images

nc: 2
names:
  0: cat
  1: dog
"""

with open("/content/data.yaml", "w") as f:
    f.write(data_yaml)

print(open("/content/data.yaml").read())
```

```python
from pathlib import Path

train_images = list(Path("/content/data/train/images").glob("*"))
val_images = list(Path("/content/data/validation/images").glob("*"))

train_labels = list(Path("/content/data/train/labels").glob("*.txt"))
val_labels = list(Path("/content/data/validation/labels").glob("*.txt"))

print("Training images :", len(train_images))
print("Training labels :", len(train_labels))
print("Validation images:", len(val_images))
print("Validation labels:", len(val_labels))
```

```python
!yolo detect train \
data=/content/data.yaml \
model=yolo11s.pt \
epochs=60 \
imgsz=640 \
batch=16 \
name=cat_dog_model
```

```python
!ls -lh /content/runs/detect/cat_dog_model/weights/
```

```python
from ultralytics import YOLO

model = YOLO("/content/runs/detect/cat_dog_model/weights/best.pt")

metrics = model.val(data="/content/data.yaml")

print(metrics)
```

```python
!yolo detect predict \
model=/content/runs/detect/cat_dog_model/weights/best.pt \
source=/content/data/validation/images \
conf=0.25 \
save=True
```
